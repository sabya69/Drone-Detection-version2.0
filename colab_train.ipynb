# 🛸 Drone ArUco Detection - CPU Fine-Tuning
Configured to run 100% reliably on standard Google Colab CPU.

### Steps:
1. In the Colab file browser (left folder icon), upload:
   - **`dataset.zip`**
   - **`best.pt`**
2. Run the cells below in order.

```python
# 1. Install Ultralytics
!pip install -q ultralytics
```

```python
# 2. Extract Dataset
!unzip -q -o dataset.zip
print("Dataset unzipped successfully!")
!ls dataset/images/train | wc -l
```

```python
# 3. Train YOLO on CPU (Optimized at 320x320 for speed)
from ultralytics import YOLO
import os

model = YOLO("best.pt")

results = model.train(
    data="dataset/dataset.yaml",
    epochs=25,
    imgsz=320,
    batch=16,
    device="cpu",
    workers=2,
    patience=10,
    project="runs/train",
    name="finetuned_drone",
    save=True,
    verbose=True
)
```

```python
# 4. Download your newly trained best.pt
from google.colab import files

best_weights = "runs/train/finetuned_drone/weights/best.pt"
if os.path.exists(best_weights):
    print(f"Downloading {best_weights} to your computer...")
    files.download(best_weights)
else:
    print("Error: could not find trained weights!")
```
